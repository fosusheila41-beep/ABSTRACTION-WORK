```python
from abc import ABC, abstractmethod

class BuildingSystem(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

    @abstractmethod
    def status(self):
        pass

class AirConditioningSystem(BuildingSystem):
    def start(self):
        print("Air Conditioning System is started")

    def stop(self):
        print("Air Conditioning System is stopped")

    def status(self):
        print("Air Conditioning System is running")

class LightingSystem(BuildingSystem):
    def start(self):
        print("Lighting System is started")

    def stop(self):
        print("Lighting System is stopped")

    def status(self):
        print("Lighting System is running")

class SecuritySystem(BuildingSystem):
    def start(self):
        print("Security System is started")

    def stop(self):
        print("Security System is stopped")

    def status(self):
        print("Security System is running")

class FireAlarmSystem(BuildingSystem):
    def start(self):
        print("Fire Alarm System is started")

    def stop(self):
        print("Fire Alarm System is stopped")

    def status(self):
        print("Fire Alarm System is running")

System = [
    AirConditioningSystem(),
    LightingSystem(),
    SecuritySystem(),
    FireAlarmSystem()
]

for system in System:
    system.start()
    system.stop()
    system.status()
```
